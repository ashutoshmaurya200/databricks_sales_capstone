```python
silver_df = spark.read.table("workspace.default.capstone_bronze_sales")
display(silver_df)
```

===============TRANSFORMATIONS=============


```python
from pyspark.sql.functions import*

silver_df = silver_df.filter(col("customer_id").isNotNull())
silver_df =silver_df.filter(col("quantity")>0)
silver_df = silver_df.filter(col("net_amount")>=0)
silver_df =silver_df.filter(col("product_id")!="UNKNOWN")

display(silver_df.count())
```

```python
silver_df =silver_df.withColumn("year" , year(col("order_date")))
silver_df =silver_df.withColumn("month" , month(col("order_date")))
silver_df =silver_df.withColumn("day" , dayofmonth(col("order_date")))
silver_df =silver_df.withColumn("month_name" , date_format(col("order_date"),"MMMM"))
display(silver_df)
```

```python
silver_df.write.mode("overwrite").saveAsTable("capstone_silver_sales")
```
