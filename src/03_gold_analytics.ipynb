```python
gold_df = spark.read.table("workspace.default.capstone_silver_sales")
display(gold_df.count())
```

==================SUMMARY FOR THE DATA=======================


```python
from pyspark.sql.functions import*

total_orders =gold_df.count()
print("Total orders: ", total_orders)

Units_sold =gold_df.agg(sum("quantity"))
print("Total units sold: ", Units_sold.collect()[0][0])

Goss_sales = gold_df.agg(sum("gross_amount"))
print("Total gross sales: ", Goss_sales.collect()[0][0])

Discount_amount = gold_df.agg(sum("discount_amount"))
print("Total discount amount: ", Discount_amount.collect()[0][0])

Net_sales = gold_df.agg(sum("net_amount"))
print("Total net sales: ", Net_sales.collect()[0][0])

avg_order_value =gold_df.agg(avg("net_amount"))
print("Average order value: ", avg_order_value.collect()[0][0])





```

```python
gold_df.write.mode("overwrite").saveAsTable("capstone_gold_sales")
```
