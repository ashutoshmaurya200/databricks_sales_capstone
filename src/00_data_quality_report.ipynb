```python


df = spark.read.table("workspace.default.capstone_bronze_sales")
display(df)
df.printSchema()
```

=========== VALIDATIONS==============

```python
data_count=df.count()
print(data_count)

from pyspark.sql.functions import*
from pyspark.sql.types import*
##CHECKING NULL RECORD OF CUSTOMER_ID

Invalid_customer_id =df.filter(col("customer_id").isNull()).count()
print(Invalid_customer_id)


##CHECkING NULL RECORD OF QUANTITY

Null_quantity =df.filter(col("quantity") <= 0).count()
print(Null_quantity)


##CHECKING INVALID RECORD OF NET AMOUNT

Null_net_amount =df.filter(col("net_amount") <0).count()
print(Null_net_amount)

##CHECKING NULL RECORD OF PRODUCT_ID

Null_product_id =df.filter(col("product_id")=="UNKNOWN").count()
print(Null_product_id)


```

CREATING FLAGS FOR INVALID DATA

```python
if Invalid_customer_id > 0 or Null_quantity>0 or Null_net_amount>0 or Null_product_id>0:
    print("Data is not valid")
else:
    print("Data is valid")
```
